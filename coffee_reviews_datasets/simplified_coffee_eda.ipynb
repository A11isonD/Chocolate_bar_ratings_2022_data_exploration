There are twelve null rows.  I am going to drop them since it's just a few rows and I have enough data otherwise.  I could fill in the missing data with the most common value (type of roast) but it might be inaccurate.  To drop the rows, I need to write dropna() and assign it to a variable.       

According to W3 Schools, "The dropna() method returns a new DataFrame object unless the inplace parameter is set to True, in that case the dropna() method does the removing in the original DataFrame instead." I originally did inplace = True before I discovered it was best to assign it to a variable.  I might have modified the original DataFrame.       

```python
df = df.dropna()
```
