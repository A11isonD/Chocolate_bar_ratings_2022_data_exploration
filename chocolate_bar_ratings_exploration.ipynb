Variables    
There are ten variables in the dataset as follows:

1. REF (reference number). The highest REF numbers were the last entries made. They are not unique values.
2. Company name or manufacturer   
3. Company location (Country)
4. Date of review of the chocolate ratings
5. Origin of bean (Country)
6. Specific bean origin or bar name
7. Cocoa percent
8. Ingredients - Represents the number of ingredients in the chocolate: (B = Beans, S = Sugar, S* = Sweetener other than white cane or beet sugar, C = Cocoa Butter, V = Vanilla, L = Lecithin, Sa = Salt)
9. Most memorable characteristics
10. Rating

Lecithin: a natural emulsifier found in egg yolks, soybeans, and sunflower seeds. It's added to chocolate to change its texture and consistency. It keeps the cocoa butter and cream from separating and improves its flow properties.

Rating Scale   
The ratings are between 1 and 5, with 1 considered the lowest rating and 5 as the highest rating possible.    

4.0 = Outstanding   
3.5 - 3.9 = Highly Recommended   
3.0 - 3.49 = Recommended   
2.0 - 2.9 = Disappointing   
1.0 - 1.9 = Unpleasant

Handling missing data  
1. Fill in the missing ingredients with the most common value.   
B, S, C (Beans, Sugar, Cocoa Butter) = 39.7%   
B, S (Beans, Sugar) = 28%  
B, S, C, L (Beans, Sugar, Cocoa Butter, Lecithin) =  11%  
These three unique value combinations together comprise 78.7% of the rows in the dataset.

2. Drop the rows with missing values. I would be dropping 3.3% of the dataset. 

3. Fill in the missing values with 'unknown' or 'not provided' for missing strings.

I will replace the missing values with the most common value (B, S, C) since I've already done visualizations on the full dataset. In the future, I would drop these rows. I decided to use the default (inplace=False) in order to return a copy where the replacing is done. I am concerned that using inplace=True to change the original DataFrame might interfere with the visualizations I've already created.

When trying to replace the NaN values in the Ingredients column, I replaced NaN with 'B, S, C' instead of '3- B,S,C' which created a new unique value. I wanted these added to the most common value, which is '3- B,S,C'. Now I will need to replace all instances of 'B, S, C' with '3- B,S,C'.  

```python
df['Ingredients'] = df['Ingredients'].replace('B, S, C', '3- B,S,C')
print(df)
```
