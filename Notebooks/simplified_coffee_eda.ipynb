I created an Entity Relationship Diagram (ERD). Now I will generate a database from it:

```python
conn = sqlite3.connect('../Data/simplified_coffee.db')

coffee_df = df[['name', 'roaster', 'roast', 'loc_country', 'origin', '100g_USD']].drop_duplicates()

coffee_rating_df = df[['rating', 'name', 'review_date', 'review']].drop_duplicates()

coffee_df.to_sql('coffee', conn, index = False, if_exists = 'replace')
coffee_rating_df.to_sql('coffee_rating', conn, index = False, if_exists = 'replace')

conn.commit
```

```python
coffee_df.to_csv('../Data/coffee.csv')
coffee_rating_df.to_csv('../Data/coffee_rating.csv')
```

I will query the coffee database.  

Which countries produce the beans for the highest-rated coffee?

```python
query1 = """
SELECT
    c.origin,
    cr.rating
FROM coffee c
JOIN coffee_rating cr ON c.name = cr.name
WHERE rating = 97 
GROUP BY c.origin
"""
```

```python
#execute the query
result1 = pd.read_sql_query(query1, conn)
result1
#This tells us the top four countries and one US state according to quality.
```

Which countries (roasters) produce the highest-rated coffee?

```python
query2 = """
SELECT
    c.loc_country,
    cr.rating
FROM coffee c
JOIN coffee_rating cr ON c.name = cr.name
WHERE rating = 97 
GROUP BY c.loc_country
"""
```

```python
result2 = pd.read_sql_query(query2, conn)
result2
```

Is a coffee's rating correlated with its price (100g in USD)? In other words, do higher-rated coffees cost more?      

If I just want to do a scatterplot to compare two numeric variables, I do not need to do any filtering or aggregation.  It will simply place each variable on a chart and I can see if there is a trend line.  Columns needed:    
1. 100g_USD
2. rating  

```python
# query goes here
query3 = """
SELECT
    c.`100g_USD`,
    cr.rating
FROM coffee c
JOIN coffee_rating cr ON c.name = cr.name
"""
# It did not like the column name starting with a number. Stack Overflow + AI overview recommended enclosing the 100g_USD in backticks or double quotes. I tried backticks first, and it worked. 
```

```python
# read query goes here
result3 = pd.read_sql_query(query3, conn)
result3
```

```python
# Assign result3 to a df variable
df_price_rating = result3
```

```python
# scatter plot goes here
plt.figure(figsize=(10, 6))
plt.scatter(df_price_rating['rating'], 
            df_price_rating['100g_USD'], 
            color = "#930B85", 
            alpha = 0.9,
            s = 18)

ax = plt.gca()
for spine in ax.spines.values():
    spine.set_linewidth(0.5)
    spine.set_alpha(0.5)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.xlabel('Rating')
plt.ylabel('Price per 100g beans In USD ')
plt.title('Price of coffee beans by rating', fontsize=15, pad=20)

plt.tight_layout()

plt.show()
```

This scatter plot shows that there is a slight upward trend in price as the rating increases.  

If I want to see the *average* price per rating, I will need to do some aggregating. 
What columns do I need for this query?    
1. 100g_USD    
2. rating    

Filtering: I don't need to filter with WHERE, because I want all of the rows.        

Sorting: I will use GROUP BY to sort the rows into groups of ratings. For example, all coffees rated 84 go together, all coffees rated 92 go together, and so on.  I will also need to get the average price of each grouping of ratings.  Finally, I will use ORDER BY to sort by rating in ascending order, even though ascending order is the default.        

```python
query4 = """
SELECT
    cr.rating,
    AVG(c.`100g_USD`) AS avg_price       
FROM coffee c
JOIN coffee_rating cr ON c.name = cr.name 
GROUP BY rating
ORDER BY rating ASC
"""
```

```python
result4 = pd.read_sql_query(query4, conn)
result4
```

I will create a bar chart with the rating on the x axis. Then I can see how the average price fluctuates, or if there is an upward/downward trend. 

```python
# Assign result4 to a df variable
df_price_rating_bar = result4
```

```python
plt.figure(figsize=(10, 6))
plt.bar(df_price_rating_bar['rating'], df_price_rating_bar['avg_price'],
        color = '#DC303E',
        width = .7
        )

plt.xlabel('Rating')
plt.ylabel('Average price per 100g beans in USD')
plt.title('Average price by coffee rating', fontsize=13, pad=15)
ax = plt.gca()
for spine in ax.spines.values():
    spine.set_linewidth(0.5)
    spine.set_alpha(0.5)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.show()
```

An interesting finding from this chart is that coffees with an 84 rating have a higher average price than those rated up to 10 points higher.  There is an otherwise general upward trend.
